## Part 5: Getting Values Out of a List and Using Them

This is one of the most important ideas in this worksheet.

When you pull a number out of a list, you can use it like a number.

````python
prices = [10, 25, 8, 40]

first_price = prices[0]

discounted_price = first_price / 2

print(discounted_price)
````

The value of `prices[0]` is `10`.

So this:

```python
prices[0] / 2
```

is really this:

```python
10 / 2
```

## Part 7: Getting Values Out of a Dictionary and Using Them

When you pull a value out of a dictionary, ask yourself:

**What type of value did I just get?**

If you pull out a number, you can do math with it.

````python
item = {
    "name": "healing potion",
    "price": 50,
    "quantity": 3,
    "is_available": True
}

total_cost = item["price"] * item["quantity"]

print(total_cost)
````

This works because:

```python
item["price"]
```

is the number `50`, and:

```python
item["quantity"]
```

is the number `3`.

So this:

```python
item["price"] * item["quantity"]
```

is really this:

```python
50 * 3
```

## Part 8: Lists of Dictionaries

A very common pattern in Python is a list of dictionaries.

````python
products = [
    {"name": "keyboard", "price": 80},
    {"name": "mouse", "price": 25},
    {"name": "monitor", "price": 200}
]

print(products[0])
print(products[0]["name"])
print(products[0]["price"])
````

Read this carefully:

```python
products[0]
```

gets the first dictionary.

```python
products[0]["name"]
```

gets the `"name"` value from the first dictionary.
